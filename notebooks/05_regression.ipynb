**IMPORT LIBRARIES**

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

**IMPORT DATA**

```python
df = pd.read_excel("../data/processed/uk_local_economy.xlsx", sheet_name='gdhi_gdp')

df.head()
```

**DROP TWO LOCAL AUTHORITIES: City of London, and Westminster. Their Values, especially their GDP per Capita, are very distorted.**

```python
distorted_local_authorities = ['City of London', 'Westminster']

df = df.loc[~df['local_authority'].isin(distorted_local_authorities), :]

df.head()
```

**SIMPLE LINEAR REGRESSION (ORDINARY LEAST SQUARES): REGRESSING GDHI per Head on GDP per Capita**

```python
df['log_gdp'] = np.log(df['2023_gdp']) # LOG-TRANSFORMATION: Standard econometric method to fix residuals, and model elasticities
df['log_gdhi'] = np.log(df['2023_gdhi'])

y_log = df['log_gdhi']
X_log = sm.add_constant(df['log_gdp'])

model_log = sm.OLS(y_log, X_log).fit()
print(model_log.summary())
```
