**IMPORT PANDAS**

```python
import pandas as pd
```

**IMPORT GDHI FILE FORM DATA/RAW**

```python
df_gdhi = pd.read_excel("../data/raw/regionalgrossdisposablehouseholdincomelocalauthorities2023.xlsx", sheet_name='Table 3', header=1)

df_gdhi.head(3)
```

**SLICE 2023 DATA**

```python
df_gdhi_23 = df_gdhi.loc[:, ['Region', 'LAD code', 'Region name', '2023']]

df_gdhi_23.head(3)
```

**RENAME COLUMNS ACCORDING TO PERSONAL STYLE**

```python
df_gdhi_23 = df_gdhi_23.rename(columns={'Region':'region', 'LAD code':'gss_ons_code', 'Region name':'local_authority'})

df_gdhi_23.head(3)
```

**VERIFY 2023 SLICE DATAFRAME FOR NANs**

```python
df_gdhi_23.isna().sum()
```

**VERIFY DATATYPES**

```python
df_gdhi_23.dtypes
```

**SORT 2023 GDHI UK LOCAL AUTHORITIES VALUES FROM HIGHEST TO LOWEST**

```python
df_gdhi_23 = df_gdhi_23.sort_values(by='2023', ascending=False)

df_gdhi_23.head(3)
```

**EXPORT TO DATA/PROCESSED**

```python
excel_path = ("../data/processed/uk_local_economy.xlsx")

with pd.ExcelWriter(
    excel_path,
    engine='openpyxl',
    mode='a',
    if_sheet_exists='replace',
) as writer:

    df_gdhi_23.to_excel(writer, sheet_name='gdhi_per_head', index=False)
```
