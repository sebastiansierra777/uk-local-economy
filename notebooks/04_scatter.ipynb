```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
```

```python
df_gdhi = pd.read_excel("../data/processed/uk_local_economy.xlsx", sheet_name='gdhi_gdp')

df_gdhi.head(3)
```

**EXCLUDE CITY OF LONDON SINCE ITS £ 8 MILLION PLUS GDP PER CAPITA IS HIGHLY DISTORTED (AND DISTORTING)** 

```python
df_gdhi = df_gdhi.loc[df_gdhi['local_authority'] != 'City of London', :]

df_gdhi.head(2)
```

```python
fig, ax = plt.subplots(figsize=(12, 9))

sns.set_theme(style='darkgrid')

sns.scatterplot(
    data=df_gdhi,
    x='2023_gdp',
    y='2023_gdhi',
    hue='region',
    alpha=0.7,
    ax=ax
)

currency_formatter = ticker.StrMethodFormatter('£{x:,.0f}')

ax.set_xscale('log')
ax.set_yscale('log')

currency_formatter = ticker.StrMethodFormatter('£{x:,.0f}')

ax.xaxis.set_major_formatter(currency_formatter)
ax.yaxis.set_major_formatter(currency_formatter)

ax.xaxis.set_minor_formatter(ticker.NullFormatter())
ax.yaxis.set_minor_formatter(ticker.NullFormatter())

ax.set_xticks([20_000, 30_000, 50_000, 100_000, 200_000, 500_000])
ax.set_yticks([20_000, 25_000, 30_000, 40_000, 50_000, 80_000, 120_000])

ax.set_title('UK Local Authorities: 2023 GDP per Capita vs 2023 GDHI per Head', fontsize=16, fontweight='bold')
ax.set_xlabel('2023 GDP per Capita', fontsize=13, fontweight='bold')
ax.set_ylabel('2023 GDHI per Head', fontsize=13, fontweight='bold')

ax.legend(loc='lower right', title='Region')

plt.show()
```
