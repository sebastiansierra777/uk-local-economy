**IMPORT PANDAS**

```python
import pandas as pd
```

**IMPORT BOTH GDP PER CAPITA AND GDHI PER HEAD SHEETS (BY UK LOCAL AUTHORITY) FROM RAW/PROCESSED FILE GENERATED WITH NOTEBOOKS '01' AND '02'**

```python
df_gdp = pd.read_excel("../data/processed/uk_local_economy.xlsx", sheet_name='gdp_per_capita_2023')

df_gdhi = pd.read_excel("../data/processed/uk_local_economy.xlsx", sheet_name='gdhi_per_head_2023')
```

**CONFIRM GDP PER CAPITA IMPORT**

```python
df_gdp.head(2)
```

**CONFIRM GDHI PER HEAD IMPORT**

```python
df_gdhi.head(2)
```

**MERGE BOTH DATAFRAMES ON 'gss_ons_code'**

```python
df_merged = pd.merge(left=df_gdp, right=df_gdhi, on='gss_ons_code', how='inner', indicator=True, suffixes=('_gdp', '_gdhi'))

df_merged.head()
```

**CONFIRM THAT THERE ARE NO NON-MERGED ENTRIES**

```python
df_merged.loc[df_merged['_merge'] != 'both']
```

**DROP UNNECESSARY COLUMNS**

```python
df_merged = df_merged.drop(columns=['ITL1_region', 'local_authority_gdhi', '_merge'])

df_merged.head(2)
```

**RENAME COLUMNS**

```python
df_merged = df_merged.rename(columns={'local_authority_gdp':'local_authority'})

df_merged.head(2)
```

**REARRANGE COLUMNS**

```python
df_merged = df_merged[['gss_ons_code', 'local_authority', 'region', '2023_gdp', '2023_gdhi']]

df_merged.head(2)
```

**CALCULATE GDHI-TO-GDP RATIO**

```python
df_merged['gdhi_gdp_ratio'] = df_merged['2023_gdhi'] / df_merged['2023_gdp']

df_merged.head(2)
```

**ROUND RATIOS TO 3 DECIMALS**

```python
df_merged['gdhi_gdp_ratio'] = df_merged['gdhi_gdp_ratio'].round(3)

df_merged.head(3)
```

**EXPORT**

```python
excel_path = ("../data/processed/uk_local_economy.xlsx")

with pd.ExcelWriter(
    excel_path,
    engine='openpyxl',
    mode='a',
    if_sheet_exists='replace',
) as writer:

    df_merged.to_excel(writer, sheet_name='gdhi_gdp', index=False)
```
